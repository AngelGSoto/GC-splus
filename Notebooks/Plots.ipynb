```python
# Script para verificar coherencia física post-homogenización
import pandas as pd
import matplotlib.pyplot as plt

# Cargar catálogo homogenizado
df = pd.read_csv('../anac_data/Results/all_fields_gc_photometry_corrected_errors_v17.csv')

# Verificar que los colores sean físicos
plt.figure(figsize=(12, 4))

# Color (F660 - F861) vs (F430 - F515) - debería mostrar secuencia
plt.subplot(131)
plt.scatter(df['MAG_F430_3'] - df['MAG_F515_3'], 
           df['MAG_F660_3'] - df['MAG_F861_3'], alpha=0.5)
plt.xlabel('F430 - F515')
plt.ylabel('F660 - F861')
plt.title('Colores S-PLUS (homogenizados)')

# Comparar con colores Taylor equivalentes
plt.subplot(132)  
plt.scatter(df['gmag'] - df['rmag'], 
           df['rmag'] - df['imag'], alpha=0.5)
plt.xlabel('g - r (Taylor)')
plt.ylabel('r - i (Taylor)')
plt.title('Colores Taylor')

plt.tight_layout()
plt.savefig('coherence_validation.png', dpi=300, bbox_inches='tight')
```
