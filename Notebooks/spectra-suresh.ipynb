```python
from astropy.io import fits
import numpy as np
from scipy.signal import find_peaks
from astropy.table import Table
import matplotlib.pyplot as plt
```

```python
# Read ASCII spectrum
spec_ascii = Table.read('../CenAGCs/mtc_189_hhh86-18.spec', format='ascii')
wavelengths_ascii = spec_ascii['col1']
flux_ascii = spec_ascii['col2']
```

```python
plt.figure(figsize=(10, 6))
plt.plot(wavelengths_ascii, flux_ascii, color='blue', linewidth=2)
plt.title('Example Spectrum')
plt.xlabel('Wavelength')
plt.ylabel('Intensity')
#plt.xlim(3000, 9000)
#plt.ylim(-0.05, 2)
plt.grid(True)
plt.show()
```
