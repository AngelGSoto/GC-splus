```python
# monitorear_analisis_cigale.py
import os
import time
import subprocess
from datetime import datetime

def monitorear_progreso():
    """Monitorea el progreso del análisis CIGALE"""
    
    print("📊 MONITOREANDO ANÁLISIS CIGALE...")
    print(f"Inicio: {datetime.now().strftime('%Y-%m-%d %H:%M:%S')}")
    
    # Verificar si existe carpeta out
    if not os.path.exists('../anac_data/CIGALE-test/out'):
        print("❌ Carpeta '../anac_data/CIGALE-test/out' no existe - ejecuta: pcigale run")
        return
    
    # Monitorear archivos en out/
    archivos_esperados = ['results.fits', 'observations.fits', 'pdfs.fits']
    
    while True:
        archivos_presentes = os.listdir('out')
        archivos_completados = [archivo for archivo in archivos_esperados if archivo in archivos_presentes]
        
        print(f"\n📁 Archivos en 'out/': {len(archivos_presentes)}")
        print(f"✅ Completados: {len(archivos_completados)}/{len(archivos_esperados)}")
        
        for archivo in archivos_esperados:
            estado = "✅" if archivo in archivos_presentes else "⏳"
            print(f"   {estado} {archivo}")
        
        if len(archivos_completados) == len(archivos_esperados):
            print("\n🎉 ANÁLISIS COMPLETADO!")
            break
        
        print("⏰ Esperando 30 segundos...")
        time.sleep(30)

def analizar_resultados_rapido():
    """Análisis rápido de los resultados una vez completados"""
    
    print("\n🔍 ANÁLISIS RÁPIDO DE RESULTADOS...")
    
    try:
        # Verificar que los archivos existen
        if not os.path.exists('out/results.fits'):
            print("❌ Resultados no disponibles aún")
            return
        
        # Análisis básico con pcigale-plots
        print("📈 Generando gráficos básicos...")
        
        # Histograma de metalicidades
        subprocess.run(['pcigale-plots', 'hist', '-p', 'stellar.metallicity'], check=True)
        
        # Histograma de edades
        subprocess.run(['pcigale-plots', 'hist', '-p', 'stellar.age_m_star'], check=True)
        
        # Histograma de masas
        subprocess.run(['pcigale-plots', 'hist', '-p', 'stellar.m_star'], check=True)
        
        print("✅ Gráficos básicos generados")
        
        # Estadísticas rápidas
        print("\n📊 ESTADÍSTICAS RÁPIDAS:")
        subprocess.run(['python', '-c', '''
from astropy.table import Table
import numpy as np
t = Table.read("../anac_data/CIGALE-test/out/results.fits")
print(f"Total objetos: {len(t)}")
print(f"Metalicidad - Mediana: {np.median(t[\"best.stellar.metallicity\"]):.4f}")
print(f"Edad - Mediana: {np.median(t[\"best.stellar.age_m_star\"]):.0f} Myr")
print(f"Masa - Mediana: {np.median(t[\"best.stellar.m_star\"]):.2e} M_sun")
'''], check=True)
        
    except Exception as e:
        print(f"⚠️  Error en análisis rápido: {e}")

def generar_graficos_publicacion():
    """Genera gráficos listos para publicación"""
    
    print("\n🎨 GENERANDO GRÁFICOS PARA PUBLICACIÓN...")
    
    # Generar SEDs para los mejores ajustes
    subprocess.run(['pcigale-plots', 'sed'], check=True)
    
    print("✅ Gráficos SED generados en ../anac_data/CIGALE-test/out/plots/sed/")
    
    # Script adicional para gráficos personalizados
    script_graficos = """
import matplotlib.pyplot as plt
from astropy.table import Table
import numpy as np

# Cargar resultados
results = Table.read('../anac_data/CIGALE-test/out/results.fits')

# Gráfico metalicidad vs edad
plt.figure(figsize=(10, 8))
plt.scatter(results['best.stellar.age_m_star'], results['best.stellar.metallicity'], 
           c=results['best.stellar.m_star'], s=50, alpha=0.7, cmap='viridis')
plt.colorbar(label='Masa estelar (M_sun)')
plt.xscale('log')
plt.xlabel('Edad (Myr)')
plt.ylabel('Metalicidad')
plt.title('Diagrama Edad-Metalicidad - Cúmulos Globulares NGC 5128')
plt.tight_layout()
plt.savefig('out/age_metallicity_diagram.png', dpi=300, bbox_inches='tight')
plt.close()

print("✅ Gráfico edad-metalicidad guardado")
"""
    
    with open('generar_graficos_personalizados.py', 'w') as f:
        f.write(script_graficos)
    
    print("📝 Script para gráficos personalizados creado: generar_graficos_personalizados.py")

if __name__ == "__main__":
    print("🔍 MONITOREO Y ANÁLISIS CIGALE - CÚMULOS GLOBULARES")
    print("=" * 70)
    
    # Preguntar qué acción realizar
    print("\n¿Qué deseas hacer?")
    print("1. Monitorear análisis en curso")
    print("2. Analizar resultados completados") 
    print("3. Generar gráficos para publicación")
    
    opcion = input("\nSelecciona opción (1-3): ").strip()
    
    if opcion == "1":
        monitorear_progreso()
    elif opcion == "2":
        analizar_resultados_rapido()
    elif opcion == "3":
        generar_graficos_publicacion()
    else:
        print("❌ Opción no válida")
```
